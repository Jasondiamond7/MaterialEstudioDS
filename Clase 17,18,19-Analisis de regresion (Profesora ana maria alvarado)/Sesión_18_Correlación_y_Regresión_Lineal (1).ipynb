# 3. Regresión Lineal Simple

```r
# 2. Ajuste del modelo de regresió lineal ----

lm_0 = lm(bio ~ 1 , data = df_bio)
summary(lm_0) #Modelo base

```

```r

## Ajustamos una regresión lineal simple:
lm_1 = lm(bio ~ ph , data = df_bio)
lm_1
```

Interpretación de parámetros:

¿Cuánto es la Biomassa para un ph de 5?

```r

summary(lm_1)


```

```r
anova(lm_1)
```

```r
df_bio |>
  ggplot(aes(x = ph, y = bio)) +
  geom_point() +
  geom_smooth(method = "lm", se = FALSE, color = "red") +
  labs(title = "Biomass vs. pH con Regresión Lineal",
       x = "pH",
       y = "Biomass")+
       theme_bw()
```

Coeficiente de Determinación

```r
summary(lm_1)$r.squared
```

# 5. Análisis de Residuos

```r
residuos_lm_1 <- residuals(lm_1)
```

```r
hist(residuos_lm_1)
```

```r
plot(lm_1)


```

## a. Normalidad


H0: Los datos siguen una distribución Normal.

H1: Los datos no siguen una distribución Normal.

```r
ggplot(data=NULL, aes(sample = residuos_lm_1))+
  stat_qq() + stat_qq_line()
```

```r
shapiro.test(residuos_lm_1)
```

```r
nortest::lillie.test(residuos_lm_1)
```

## b. Independencia

H0: Los errores no están autocorrelacionados.

H1: Los errores presentan correlación a un paso.

```r

lmtest::dwtest(lm_1)

```

## c. Homocedasticidad

H0: Los errores son homocedásticos

H1: Los errores no son heterocedásticos

```r
lmtest::bptest(lm_1)



```
