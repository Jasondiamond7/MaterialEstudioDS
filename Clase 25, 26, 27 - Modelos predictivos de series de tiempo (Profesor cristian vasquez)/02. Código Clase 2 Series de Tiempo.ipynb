```r
# ----------------------------
# AR(1)
# ----------------------------

set.seed(1234)

AR1 <- arima.sim(list(order = c(1,0,0), ar = 0.8), n = 200)

ggAcf(AR1,col="#0072B2",type="correlation")    +
    ylim(-0.2,1)                               +
    ggtitle("ACF de AR(1)")                    +
    theme_minimal()
```

```r
##############################
### ACF Y PACF AR(1)
##############################

set.seed(1234)

AR1 <- arima.sim(list(order = c(1,0,0), ar = 0.8), n = 200)

plot1 = ggAcf(AR1,col="#0072B2",type="correlation")    +
            ylim(-0.2,1)                               +
            ggtitle("ACF de AR(1)")                    +
            theme_minimal()

plot2 = ggPacf(AR1,col="#0072B2",type="correlation")    +
            ylim(-0.2,1)                                +
            ggtitle("PACF de AR(1)")                    +
            theme_minimal()

grid.arrange(plot1, plot2, ncol=2)
```

```r
##############################
### ACF Y PACF AR(2)
##############################

set.seed(123)

AR2 <- arima.sim(list(order = c(2,0,0), ar = c(0.5, 0.3)), n = 600)

plot1 = ggAcf(AR2,col="#0072B2",type="correlation")    +
            ylim(-0.2,1)                               +
            ggtitle("ACF de AR(2)")                    +
            theme_minimal()

plot2 = ggPacf(AR2,col="#0072B2",type="correlation")    +
            ylim(-0.2,1)                                +
            ggtitle("PACF de AR(2)")                    +
            theme_minimal()

grid.arrange(plot1, plot2, ncol=2)
```

# Ejemplo

En el archivo **tree.xlsx** se encuentra la medición del crecimiento anual de los anillos de un árbol en Idaho, USA. Estas series son de mucha utilidad en el área de la climatología ya que se puede reconstruir el clima, precipitación, temperatura del pasado en períodos de los que no se tiene constancia meteorológica de ningún tipo.

```r
##############################
### Ajuste de modelos ARMA
##############################


tree = read_xlsx("tree.xlsx", sheet = 1, col_types = c("numeric","numeric"))
tree
```

```r
##############################
### Gráfico de la serie
##############################

ggplot(tree, aes(x = Year, y = Idaho)) 		  +
    geom_line(color = "#0072B2",size=1.1)	 	+
    xlab("Date"	)                          	+
    ylab("Ancho anuales en CM")             +
    theme_minimal()
```

```r
mean(tree$Idaho)
median(tree$Idaho)
sd(tree$Idaho)
var(tree$Idaho)
IQR(tree$Idaho)
summary(tree$Idaho)
```

```r
plot1 = ggAcf(tree$Idaho,col="#0072B2",type="correlation", lag.max = 15)     	+
            ylim(-0.2,0.75)                             				              +
            ggtitle("ACF ")		  	              				                      +
            theme_minimal()

plot2 = ggPacf(tree$Idaho,col="#0072B2",type="correlation", lag.max = 15)     +
            ylim(-0.2,0.75)                             				              +
            ggtitle("PACF ")		  	              				                    +
            theme_minimal()

grid.arrange(plot1, plot2, ncol=2)
```

```r
# ----------------------------------------------------------------------------
# Ajustes diferentes modelos
# ----------------------------------------------------------------------------


### librería tseries

modelo1 = arma(tree$Idaho, order = c(0,3))
summary(modelo1)
```

```r
modelo2 = arma(tree$Idaho, order = c(0,2))
summary(modelo2)
```

```r
modelo3 = arma(tree$Idaho, order = c(1,0))
summary(modelo3)
```

```r
modelo4 = arma(tree$Idaho, order = c(2,0))
summary(modelo4)
```

```r
modelo5 = arma(tree$Idaho, order = c(1,1))
summary(modelo5)
```

```r
### librería forecast

Arima(tree$Idaho,order = c(0,0,3))
Arima(tree$Idaho,order = c(0,0,2))
Arima(tree$Idaho,order = c(1,0,0))
Arima(tree$Idaho,order = c(2,0,0))
Arima(tree$Idaho,order = c(1,0,1))
```

```r
##############################
### Valores Ajustados
##############################

tree$Ajuste = fitted(modelo5)

colors <- c("Idaho" = "blue", "Ajuste" = "red")

ggplot(tree, aes(x = Year))                                 +
    geom_line(aes(y = Idaho, color = "Idaho"),size=1.1)	 	  +
    geom_line(aes(y = Ajuste, color = "Ajuste"),size=1.1)	 	+
    xlab("Year")                          				          +
    ylab("Ancho anuales en CM")          					          +
    ggtitle("Serie Climatología")						                +
    scale_y_continuous(labels = number)					            +
    scale_color_manual(values = colors)					            +
    labs(color = "Legend") 							                    +
    theme_minimal()
```

```r
ggtsdiag(Arima(tree$Idaho,order = c(1,0,1)), gof.lag = 20,ad.colour="blue")	+
  theme_minimal()
```

```r
##############################
### Residuos
##############################


tree$Residuos = residuals(modelo5)

plot1 <- ggplot(tree, aes(Residuos))   	   +
            geom_histogram(alpha = 0.5,
                          fill  = "blue",
                          col   = "red",
                          bins = 20)	     +
            theme_minimal()


plot2 <- ggplot(tree, aes(sample = Residuos)) 	+
            stat_qq(color = "blue") 		        +
            stat_qq_line(color="red")		        +
            theme_minimal()

grid.arrange(plot1, plot2, ncol=2)
```

```r
media = mean(tree$Residuos,na.rm = TRUE)
desv  = sd(tree$Residuos,na.rm = TRUE)
ks.test(tree$Residuos, "pnorm",media,desv)
```

```r
Residuos     <- tree$Residuos[!is.na(tree$Residuos)]
distribucion <-fitdist(Residuos, "norm")

summary(distribucion)
plot(distribucion)
```

```r
##############################
### Predicción
##############################

forecast1 <-forecast(Arima(tree$Idaho,order = c(1,0,1)), level = c(95), h = 10)
autoplot(forecast1)
```
