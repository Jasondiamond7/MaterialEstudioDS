```r
options(contrasts = c("contr.treatment", "contr.treatment")) # elimina los contrastes polinomiales
mod_mult <- multinom(PID3 ~ age + educ + income, data = nes96)
summary(mod_mult)
```

```r
coeficientes <- summary(mod_mult)$coefficients
errores <- summary(mod_mult)$standard.errors

coeficientes
errores
```

```r
prob_hat <- fitted(mod_mult)

head(prob_hat)
```

```r
pred_clase <- predict(mod_mult, type = "class")

table(Observado = nes96$PID3, Predicho = pred_clase)
```

```r
#install.packages("caret")
library(caret)

# Asegurarse que observado y predicho tengan los mismos niveles
pred_clase <- factor(pred_clase, levels = levels(nes96$PID3))

# Matriz de confusión y medidas
confusionMatrix(
  data = pred_clase,
  reference = nes96$PID3
)
```

```r
mod_null <- multinom(PID3 ~ 1, data = nes96)
```

```r
logLik(mod_null)
logLik(mod_mult)
```

```r
LR <- 2 * (as.numeric(logLik(mod_mult)) - as.numeric(logLik(mod_null)))
LR
```

```r
gl <- attr(logLik(mod_mult), "df") - attr(logLik(mod_null), "df")
gl
```

```r
pvalor <- pchisq(LR, df = gl, lower.tail = FALSE)
pvalor
```

```r
if(pvalor < 0.05){
  "El modelo con covariables mejora significativamente respecto al modelo nulo."
} else {
  "No hay evidencia suficiente de que las covariables mejoren el modelo nulo."
}
```

```r
deviance(mod_mult)
deviance(mod_null)
```

```r
deviance(mod_null) - deviance(mod_mult)
```
